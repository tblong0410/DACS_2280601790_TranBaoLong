```python
from pathlib import Path
import json
import pandas as pd

# Thư mục dữ liệu EMBER
DATA_DIR = Path(r"D:\DACS\ember2018")

# Lấy 50 mẫu có label = 1
rows = []

for i in range(6):
    file_path = DATA_DIR / f"train_features_{i}.jsonl"

    print("Đang đọc:", file_path.name)

    with file_path.open("r", encoding="utf-8") as f:
        for line in f:
            if not line.strip():
                continue

            record = json.loads(line)

            if record.get("label") == 1:
                rows.append(record)

                if len(rows) == 50:
                    break

    if len(rows) == 50:
        break

# Chuyển thành bảng DataFrame
df_malware = pd.DataFrame(rows)

print("\nSố mẫu lấy được:", len(df_malware))
print("Số cột:", len(df_malware.columns))
print("\nTên các cột:")
print(df_malware.columns.tolist())
```

```python
import json
import pandas as pd

def compact_value(value):
    if isinstance(value, dict):
        return json.dumps(
            value, ensure_ascii=False
        )[:150] + ("..." if len(json.dumps(value, ensure_ascii=False)) > 150 else "")

    if isinstance(value, list):
        return f"List({len(value)} phần tử): {value[:5]}"

    return value

df_view = df_malware.copy()

for column in df_view.columns:
    df_view[column] = df_view[column].map(compact_value)

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 200)

display(df_view)
```
